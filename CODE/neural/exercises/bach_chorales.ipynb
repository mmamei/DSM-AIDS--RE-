It is composed of 382 chorales composed by Johann Sebastian Bach. Each chorale is 100 to 640 time steps long, and each time step contains 4 integers, where each integer corresponds to a note's index on a piano (except for the value 0, which means that no note is played). Train a model—recurrent, convolutional, or both—that can predict the next time step (four notes), given a sequence of time steps from a chorale. Then use this model to generate Bach-like music, one note at a time: you can do this by giving the model the start of a chorale and asking it to predict the next time step, then appending these time steps to the input sequence and asking the model for the next note, and so on.

```python
import os
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import keras
```

```python
import os

def read_files(dir):
    files = []
    for file in os.listdir(dir):
        df = pd.read_csv(dir + '/' + file)
        files.append(df.values.tolist())
    return files


train_chorales = read_files('../../datasets/jsb_chorales/train')
valid_chorales = read_files('../../datasets/jsb_chorales/valid')
test_chorales = read_files('../../datasets/jsb_chorales/test')

train_chorales[0][:5]
```

```python
notes = set()
for chorales in (train_chorales, valid_chorales, test_chorales):
    for chorale in chorales:
        for chord in chorale:
            notes |= set(chord)

n_notes = len(notes)
min_note = min(notes - {0})
max_note = max(notes)
print('n_notes:', n_notes)
print('min_note:', min_note)
print('max_note:', max_note)
```

Notes range from 36 (C1 = C on octave 1) to 81 (A5 = A on octave 5), plus 0 for silence

```python
from IPython.display import Audio

def notes_to_frequencies(notes):
    # Frequency doubles when you go up one octave; there are 12 semi-tones
    # per octave; Note A on octave 4 is 440 Hz, and it is note number 69.
    return 2 ** ((np.array(notes) - 69) / 12) * 440

def frequencies_to_samples(frequencies, tempo, sample_rate):
    note_duration = 60 / tempo # the tempo is measured in beats per minutes
    # To reduce click sound at every beat, we round the frequencies to try to
    # get the samples close to zero at the end of each note.
    frequencies = np.round(note_duration * frequencies) / note_duration
    n_samples = int(note_duration * sample_rate)
    time = np.linspace(0, note_duration, n_samples)
    sine_waves = np.sin(2 * np.pi * frequencies.reshape(-1, 1) * time)
    # Removing all notes with frequencies ≤ 9 Hz (includes note 0 = silence)
    sine_waves *= (frequencies > 9.).reshape(-1, 1)
    return sine_waves.reshape(-1)

def chords_to_samples(chords, tempo, sample_rate):
    freqs = notes_to_frequencies(chords)
    freqs = np.r_[freqs, freqs[-1:]] # make last note a bit longer
    merged = np.mean([frequencies_to_samples(melody, tempo, sample_rate)
                     for melody in freqs.T], axis=0)
    n_fade_out_samples = sample_rate * 60 // tempo # fade out last note
    fade_out = np.linspace(1., 0., n_fade_out_samples)**2
    merged[-n_fade_out_samples:] *= fade_out
    return merged

def play_chords(chords, tempo=160, amplitude=0.1, sample_rate=44100, filepath=None):
    samples = amplitude * chords_to_samples(chords, tempo, sample_rate)
    if filepath:
        from scipy.io import wavfile
        samples = (2**15 * samples).astype(np.int16)
        wavfile.write(filepath, sample_rate, samples)
        return display(Audio(filepath))
    else:
        return display(Audio(samples, rate=sample_rate))
```

```python
# Lets create a function to flatten a chorale 
def flatten_chorale(chorale):
    return [note - 35 if note > 0 else 0 for chord in chorale for note in chord]

# Lets create a function to get back the original chorale from the flattened one
def unflatten_chorale(chorale, n_chords):
    tmp = [note + 35 if note > 0 else 0 for note in chorale]
    return [tmp[i:i + n_chords] for i in range(0, len(chorale), n_chords)]

flat = flatten_chorale(train_chorales[10])
unflat = unflatten_chorale(flat, 4)

#play_chords(unflat)
```

### Now we can consider the chorale as a time-series and predict/create new chorales

```python
# create the dataset

train_chorales = [flatten_chorale(c) for c in train_chorales]
valid_chorales = [flatten_chorale(c) for c in valid_chorales]
test_chorales = [flatten_chorale(c) for c in test_chorales]

seq_length = 32
X_train = []
y_train = []
for c in train_chorales:
    for i in range(len(c)-seq_length-1):
        X_train.append(c[i:i+seq_length])
        y_train.append(c[i+1:i+seq_length+1])

X_train = np.array(X_train)
y_train = np.array(y_train)
print(X_train.shape)
print(y_train.shape)
```

```python
model = keras.models.Sequential([
    keras.layers.Input(shape=[None, 1]),
    keras.layers.LSTM(20, return_sequences=True),
    #keras.layers.LSTM(20, return_sequences=True),
    keras.layers.TimeDistributed(keras.layers.Dense(n_notes, activation="softmax"))
])
model.summary()
```

```python
callbacks = [
    keras.callbacks.EarlyStopping(monitor="val_loss", patience=5, verbose=1),
]

model.compile(loss="sparse_categorical_crossentropy", optimizer="adam")
history = model.fit(X_train, y_train, epochs=20, validation_split=0.2, callbacks=callbacks)
model.save('saved_bach.keras')
```

```python
import tensorflow as tf
def next_char(X_new, temperature=1):
    X_new = [note - 35 if note > 0 else 0 for note in X_new]
    X_new = np.array(X_new).reshape(1, -1, 1)
    y_proba = model.predict(X_new)[0, -1:, :]
    rescaled_logits = tf.math.log(y_proba) / temperature
    char_id = tf.random.categorical(rescaled_logits, num_samples=1) 
    next_note = (char_id.numpy()[0]) # this is the predicted next character
    next_note = next_note + 35 if next_note > 0 else 0
    return next_note    


'''
def complete_text(text, n_chars=100, temperature=1):
    for _ in range(n_chars):
        text += next_char(text, temperature)
    return text

#out = complete_text("First Citizen:", temperature=0.1) 
out = complete_text("To be or not to be", temperature=0.4) 
print(out)
'''
next_char([65,60,57,53])
```
