# Example 1

The average time to manufacture an item if 100 s, with a standard deviation of 10 s.
A company invents a special manufacturing procedure to produce items at least 5 s faster. 

The manager analyzes a sample of the production of 20 items. The average production time is 94 s. Is there sufficient evidence to support her thesis?


```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
```

```python


# Known population parameters
mu_0 = 95       # threshold under H0: new procedure is at most 5s faster (100-5)
sigma = 10      # population std deviation (known)
n = 100          # sample size
x_bar = 93      # sample mean

# H0: mu >= 95  (improvement < 5s)
# H1: mu < 95   (improvement >= 5s, left-tailed)
alpha = 0.05

# Z statistic
z = np.sqrt(n) * (x_bar - mu_0) / sigma

# Critical value and p-value (left-tailed)
z_critical = stats.norm.ppf(alpha) 
p_value = stats.norm.cdf(z)

print(f"Z statistic:      {z:.4f}")
print(f"Critical value:   {z_critical:.4f}  (alpha={alpha})")
print(f"p-value:          {p_value:.4f}")
print()
if z < z_critical:
    print("REJECT H0: sufficient evidence the procedure is at least 5s faster.")
else:
    print("FAIL TO REJECT H0: insufficient evidence the procedure is at least 5s faster.")
```

# Example 2

The average weekly sales of a company is 1M, with a standard deviation of 50K.
A new advertising campaign aims at increasing sales by 10%.

The manager analyzes sales data over 10 weeks. The average weekly sales is 1.2M. Is there sufficient evidence to support her thesis?





```python
# Known population parameters
mu_0 = 1_000_000    # baseline weekly sales
sigma = 50_000      # population std deviation (known)
n = 10              # sample size (weeks)
x_bar = 1_200_000   # sample mean

# H0: mu <= 1.1M  (increase < 10%)
# H1: mu > 1.1M   (increase >= 10%, right-tailed)
mu_threshold = mu_0 * 1.10
alpha = 0.05

# Z statistic
z = np.sqrt(n) * (x_bar - mu_threshold) / sigma

# Critical value and p-value (right-tailed)
z_critical = stats.norm.ppf(1 - alpha)
p_value = 1 - stats.norm.cdf(z)

print(f"Z statistic:      {z:.4f}")
print(f"Critical value:   {z_critical:.4f}  (alpha={alpha})")
print(f"p-value:          {p_value:.4f}")
print()
if z > z_critical:
    print("REJECT H0: sufficient evidence sales increased by at least 10%.")
else:
    print("FAIL TO REJECT H0: insufficient evidence sales increased by at least 10%.")
```

# Example 3

A company manufactures cylindrical metal components with a target average diameter of 5 mm and standard deviation of 0.1 mm.
The quality-control engineer wants to verify whether the manufacturing process is still correctly centered on the target value. A sample of produced components is measured.

The manager analyzes a sample of the production of 20 items. The average diameter is 5.01 mm. Is there sufficient evidence to support his thesis?

```python
mu_0 = 5.0      # target diameter
sigma = 0.1
n = 20
x_bar = 5.01

# H0: mu = 5  (process centered)
# H1: mu != 5 (two-tailed)
alpha = 0.05

z = np.sqrt(n) * (x_bar - mu_0) / sigma

p_value = 2 * (1 - stats.norm.cdf(abs(z)))   # two-tailed

print(f"Z statistic:  {z:.4f}")
print(f"p-value:      {p_value:.4f}")
print()
if p_value < alpha:
    print("REJECT H0: sufficient evidence the process is off-center.")
else:
    print("FAIL TO REJECT H0: insufficient evidence the process is off-center.")
```

```python

# create a sample of 100 observations from a normal distribution with mean 94 and std deviation 10

np.random.seed(42)
sample = np.random.normal(94, 10, 100)  
print(f"Sample mean: {np.mean(sample):.4f}")

from statsmodels.stats.weightstats import ztest
# left-tailed
z, p = ztest(sample, value=95, alternative='smaller')
print(f"Z statistic: {z:.4f}, p-value: {p:.4f}")


```
