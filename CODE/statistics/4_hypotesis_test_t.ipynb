# Example 1

The average time to manufacture an item if 100 s.
A company invents a special manufacturing procedure to produce items at least 5 s faster. 

The manager analyzes a sample of the production of 15 items. The average production time is 94 s, Estimated standard deviation s = 10s. Is there sufficient evidence to support her thesis?


```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
```

```python


# Known population parameters
mu_0 = 95       # threshold under H0: new procedure is at most 5s faster (100-5)

n = 15          # sample size
x_bar = 93      # sample mean
s = 10          # sample  std deviation

# H0: mu >= 95  (improvement < 5s)
# H1: mu < 95   (improvement >= 5s, left-tailed)
alpha = 0.05

# t statistic
t = (x_bar - mu_0) / (s / np.sqrt(n))
print(f"t statistic: {t:.4f}")

# compute p-value for left-tailed test
p_value = stats.t.cdf(t, df=n-1)
print(f"p-value: {p_value:.4f}")

# compute critical t value for left-tailed test
t_critical = stats.t.ppf(alpha, df=n-1)
print(f"t critical value: {t_critical:.4f}")

if p_value < alpha:
    print("Reject H0: the new procedure is significantly faster than the old one.")
else:
    print("Fail to reject H0: there is not enough evidence to conclude that the new procedure is faster than the old one.")

```

# Example 1 with Data




```python

# create a sample of 15 observations from a normal distribution with mean 93 and std deviation 10

np.random.seed(42)
sample = np.random.normal(93, 10, 15)  
print(f"Sample mean: {np.mean(sample):.4f}")
print(f"Sample std deviation: {np.std(sample, ddof=1):.4f}")

# compute t-test using scipy.stats.ttest_1samp
t_statistic, p_value = stats.ttest_1samp(sample, mu_0)
print(f"t statistic: {t_statistic:.4f}")
print(f"p-value: {p_value/2:.4f}")  # divide by


```
