# CDF and PPF



```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
```

```python
p = 0.05  # significance level
z = stats.norm.ppf(p)
print(z)
p_value = stats.norm.cdf(z)
print(p)
```

```python
# draw a plot to show the relation between the critical value and the p-value

probabilities = np.linspace(0, 1, 1000)
z_values = stats.norm.ppf(probabilities)
plt.plot(z_values, probabilities)
plt.xlabel('Z Values')
plt.ylabel('Cumulative Probability')
plt.axvline(x=z, color='r', linestyle='--', label=f'Z Value (z={z:.2f})')
plt.axhline(y=p, color='g', linestyle='--', label=f'Probability (p={p:.2f})')
plt.legend()
plt.show()
```

```python
# what is the probability between -sigma and +sigma for a normal distribution?

p = stats.norm.cdf(2) - stats.norm.cdf(-2)
print(p)  
```

```python
# what is the area that corresponds to the 95%?

z1 = stats.norm.ppf(0.025)
print(z1)
z2 = stats.norm.ppf(0.975)
print(z2)
p = stats.norm.cdf(z2) - stats.norm.cdf(z1)
print(p)
```
