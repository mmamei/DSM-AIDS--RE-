```python
import math
from statsmodels.stats.power import normal_sample_size_one_tail

mu_0 = 100
mu_true = 95
sigma = 10
alpha = 0.05
target_power = 0.90

difference = abs(mu_true - mu_0)

n = normal_sample_size_one_tail(
    diff=difference,
    power=target_power,
    alpha=alpha,
    std_null=sigma
)

print(f"Unrounded sample size: {n:.3f}")
print(f"Required sample size: {math.ceil(n)}")
```
