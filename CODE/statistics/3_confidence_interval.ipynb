# Example 1

The average time to manufacture an item if 100 s, with a standard deviation of 10 s.
A company invents a special manufacturing procedure to produce items at least 5 s faster. 

The manager analyzes a sample of the production of 20 items. The average production time is 94 s.

```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
```

```python


# Known population parameters
mu_0 = 95       # threshold under H0: new procedure is at most 5s faster (100-5)
sigma = 10      # population std deviation (known)
n = 100          # sample size
x_bar = 93      # sample mean

# H0: mu >= 95  (improvement < 5s)
# H1: mu < 95   (improvement >= 5s, left-tailed)
alpha = 0.05

# Z statistic
z = np.sqrt(n) * (x_bar - mu_0) / sigma

# Critical value and p-value (left-tailed)
z_critical = stats.norm.ppf(alpha) 
p_value = stats.norm.cdf(z)
```

```python
# confidence interval for the mean

ci = (x_bar - z_critical * (sigma / np.sqrt(n)), x_bar + z_critical * (sigma / np.sqrt(n)))
ci_lower, ci_upper = sorted(ci)
print(ci_lower, ci_upper)
```

```python
# create a bar plot with the x_bar and confidence interval
plt.bar(1, x_bar, color='blue', alpha=0.7, width=0.2, label='Sample Mean')
plt.errorbar(1,x_bar,yerr=[[x_bar - ci_lower], [ci_upper - x_bar]], color='black', capsize=5, label='Confidence Interval')
plt.xlim(0, 2)
plt.show()
```

```python
import seaborn as sns

np.random.seed(42)
sample = np.random.normal(94, 10, 100)  
print(f"Sample mean: {np.mean(sample):.4f}")
sns.barplot(x=np.ones(len(sample)), y=sample, capsize=0.1, width=0.2)
plt.xlim(-1, 1)
plt.show()
```
