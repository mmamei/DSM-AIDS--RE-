```python

# voglio usare sklleran per creare un dataset di esempio a blobs
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt
import numpy as np
# creo un dataset di esempio con 300 campioni, 3 centri e 2 caratteristiche
X, y = make_blobs(n_samples=300, centers=3, cluster_std=0.60, random_state=0)
# visualizzo il dataset
plt.scatter(X[:, 0], X[:, 1], s=50)
plt.show()
```

```python
# moltiplico la prima colonna per 100
X[:, 0] = X[:, 0] * 10000
```

```python
plt.scatter(X[:, 0], X[:, 1], s=50)
plt.show()
```

```python
# voglio un dataset con un cluster di punti al centro e un cerchio di punti intorno
from sklearn.datasets import make_circles
X, y = make_circles(n_samples=300, factor=0.05, noise=0.05)
plt.scatter(X[:, 0], X[:, 1], s=50)
plt.show()
```

```python
# voglio aggiungere una x[:,3] che sia la somma di x[:,0] e x[:,1] al quadrato
X = np.hstack((X, (X[:, 0:1]**2 + X[:, 1:2]**2)))
```

```python
X
```

```python
# min max scaling di X
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
X = scaler.fit_transform(X)
```

```python
X
```

```python
# voglio usare kmeans per fare il clustering del dataset di esempio
from sklearn.cluster import KMeans
kmeans = KMeans(n_clusters=2)
kmeans.fit(X)
y_kmeans = kmeans.predict(X)
# visualizzo i risultati del clustering
plt.scatter(X[:, 0], X[:, 1], c=y_kmeans, s=50, cmap='viridis')
centers = kmeans.cluster_centers_   
plt.scatter(centers[:, 0], centers[:, 1], c='red', s=200, alpha=0.75)
plt.show()
```

```python
# crea un dataframe con i dati di X e le etichette di y_kmeans
import pandas as pd
df = pd.DataFrame(X, columns=['Feature 1', 'Feature 2'])
df['Cluster'] = y_kmeans    
df.head()
```
