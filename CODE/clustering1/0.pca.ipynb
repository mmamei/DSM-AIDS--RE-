```python
import math
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = [np.random.uniform(0,50) for i in range(1000) ]
y = [np.random.uniform(0,50) for i in range(1000) ]
data = np.array( list( zip(x,y) ) )
plt.axis('equal')
#plt.scatter(x, y)
#plt.show()
plt.scatter(data[:,0], data[:,1])
plt.show()
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=1)
pca.fit(data)
data_pca = pca.transform(data)
```

```python

plt.scatter(data_pca[:,0], data_pca[:,1])
# make aspect ratio equal
plt.axis('equal')
plt.show()
```

```python
# print eigenvectors and eigenvalues
print(pca.components_)
print(pca.explained_variance_)
```

```python
reconstructed_data = pca.inverse_transform(data_pca)
plt.scatter(reconstructed_data[:,0], reconstructed_data[:,1])
plt.axis('equal')
plt.show()
```
