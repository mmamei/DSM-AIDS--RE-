```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('01 - SuperStoreUS-2015.csv', sep=';')
df.head(2)
```

```python
# voglio creare una colonna numerica che rappresenta product category.
# deve avere i valori 1,2,3 associati a Furniture, Office Supplies, Technology
df['Product Category Numeric'] = df['Product Category'].map({'Furniture': 1, 'Office Supplies': 2, 'Technology': 3})  
df.head(2)  
```

```python
plt.scatter(df['Product Category Numeric'], df['Sales'], alpha=0.5)
plt.xlabel('Product Category')
plt.ylabel('Sales')
plt.title('Sales by Product Category')
plt.xticks([1, 2, 3], ['Furniture', 'Office Supplies', 'Technology'])
plt.show()  
```

```python
df.columns
```

```python
plt.scatter(df['Product Category Numeric'], df['Quantity ordered new'], alpha=0.5)
plt.xlabel('Product Category')
plt.ylabel('Quantity Ordered')
plt.title('Quantity Ordered by Product Category')
plt.xticks([1, 2, 3], ['Furniture', 'Office Supplies', 'Technology'])
plt.show() 
```

```python
# voglio uno sctter tra 'Quantity ordered new' e 'Sales' solo per i
# prodotti 'Furniture'
furniture_df = df[df['Product Category'] == 'Furniture']
plt.scatter(furniture_df['Quantity ordered new'], furniture_df['Sales'], alpha=0.5)
plt.xlabel('Quantity Ordered')
plt.ylabel('Sales')
plt.title('Sales vs Quantity Ordered for Furniture')
plt.show()
```

```python
furniture_df = df[df['Product Category'] == 'Furniture']
m = max(furniture_df['Sales'])
furniture_df = furniture_df[furniture_df['Sales'] < m]
```

```python
plt.scatter(furniture_df['Quantity ordered new'], furniture_df['Sales'], alpha=0.5)
plt.xlabel('Quantity Ordered')
plt.ylabel('Sales')
plt.title('Sales vs Quantity Ordered for Furniture')
plt.show()
```
