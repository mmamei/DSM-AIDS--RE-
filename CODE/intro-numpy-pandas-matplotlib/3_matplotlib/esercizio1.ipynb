```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../../dataset/autoMpg.csv', na_values=['?'])
df.head(2)
```

```python
m = max(df['mpg'])
df.loc[df['mpg'] == m,'mpg'] = 27.0
```

```python
plt.scatter(df['displacement'],df['mpg'])
plt.xlabel('displacement')
plt.ylabel('mpg')
plt.title('displacement vs mpg')
plt.show()
```

```python
plt.scatter(df['horsepower'],df['mpg'])
plt.xlabel('horsepower')
plt.ylabel('mpg')
plt.title('horsepower vs mpg')
plt.show()
```

```python
plt.scatter(df['cylinders'],df['mpg'])
plt.xlabel('cylinders')
plt.ylabel('mpg')
plt.title('cylinders vs mpg')
plt.show()
```

```python
# voglio vedere uno scatter tra horsepower e mpg, ma voglio considerare
# solo le auto con 4 cilindri
df4 = df[df['cylinders'] == 4]
df8 = df[df['cylinders'] == 8]
df6 = df[df['cylinders'] == 6]
plt.scatter(df4['horsepower'],df4['mpg'], label='4 cylinders')
plt.scatter(df8['horsepower'],df8['mpg'], label='8 cylinders')
plt.scatter(df6['horsepower'],df6['mpg'], label='6 cylinders')  
plt.xlabel('horsepower')
plt.ylabel('mpg')
plt.title('horsepower vs mpg')
plt.legend(loc='best')    
plt.show()
```
